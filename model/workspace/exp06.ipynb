```python
from yolov3.models.yolo import (
    BaseModel,
)  # notebookの場合パスが通らないのでこのようにする
from src.application.use_case.train_model import TrainModel
from torchvision import transforms
```

```python
config_dir = "../config/"
config_path = (
    config_dir + "model_architecture/omr_yolov5_change_anchor.yaml"
)  #'yolov3/models/yolov5s.yaml'
hyp_path = config_dir + "hyp/hyp.scratch-low.yaml"
model_dir = "../models/"
model_path = model_dir + "pre_trained/yolov5s.pt"

data_dir = "../data/"
img_dir = data_dir + "dense/images/"
annotation_dir = data_dir + "dense/labels_mapping_to_center_and_normalization/"


output_path = model_dir + "fine_tuned/omr_yolov5s_change_anchor_finetuned.pth"
ratio_wh = 1
RESIZE_SIZE_W = 800
RESIZE_SIZE_H = int(ratio_wh * RESIZE_SIZE_W)
print(RESIZE_SIZE_H, RESIZE_SIZE_W)

epochs = 3
batch_size = 8
```

```python
# モデル学習
transform = transforms.Compose(
    [
        transforms.Resize((RESIZE_SIZE_H, RESIZE_SIZE_W)),
        transforms.ToTensor(),
    ]
)
train_model = TrainModel(
    config_path,
    model_path,
    hyp_path,
    img_dir,
    annotation_dir,
    output_path,
    epochs=epochs,
    batch_size=batch_size,
    transform=transform,
)
train_model.train()
```
