```python
import os
import cv2
from PIL import Image
from pathlib import Path
```

```python
def draw_bboxes(image_path, annotations, output_size: int = 0):
    """
    image_path: 入力画像ファイルへのパス (str)
    annotations: [[cx_ratio, cy_ratio, w_ratio, h_ratio, class_label], ...]
    """
    # 画像の読み込み
    img = cv2.imread(image_path)
    if img is None:
        raise FileNotFoundError(f"Image not found at {image_path}")

    if output_size != 0:
        img = cv2.resize(img, (output_size, output_size), interpolation=cv2.INTER_AREA)

    h, w = img.shape[:2]

    for ann in annotations:
        cx_ratio, cy_ratio, w_ratio, h_ratio, class_label = ann

        # 実ピクセル座標へ変換
        cx = cx_ratio * w
        cy = cy_ratio * h
        bw = w_ratio * w
        bh = h_ratio * h

        # バウンディングボックスの左上・右下座標を計算
        x1 = int(cx - bw / 2)
        y1 = int(cy - bh / 2)
        x2 = int(cx + bw / 2)
        y2 = int(cy + bh / 2)
        # print(x1, y1, x2, y2)

        # バウンディングボックスの描画
        # (B, G, R) 例: 緑色
        cv2.rectangle(img, (x1, y1), (x2, y2), (0, 255, 0), 2)

        # クラスラベルを表示（任意でフォントやサイズを調整可）
        label_str = str(class_label)
        font = cv2.FONT_HERSHEY_SIMPLEX
        font_scale = 0.6
        thickness = 1

        # テキストサイズ取得し、背景ボックスを描画してからテキストを重ねる
        (text_w, text_h), _ = cv2.getTextSize(label_str, font, font_scale, thickness)
        cv2.rectangle(img, (x1, y1 - text_h - 5), (x1 + text_w, y1), (0, 255, 0), -1)
        cv2.putText(
            img, label_str, (x1, y1 - 5), font, font_scale, (0, 0, 0), thickness
        )

    return img


def load_annotations(annotation_path):
    with open(annotation_path, "r") as file:
        lines = file.readlines()

    annotations = []
    for line in lines:
        class_label, pitch_label, cx_ratio, cy_ratio, w_ratio, h_ratio = line.split()
        annotations.append(
            [
                float(cx_ratio),
                float(cy_ratio),
                float(w_ratio),
                float(h_ratio),
                int(class_label),
            ]
        )
    return annotations
```

```python
image_number = 1
OUTPUT_IMG_SIZE = 1280
data_dir = Path("../data/complete")
output_path = Path("../data/output/complete/true")
image_path = data_dir / f"images/{image_number}.png"
im0 = cv2.imread(image_path.as_posix())  # annotaterはこっちを使っている
image = Image.open(image_path).convert("RGB")

annotation_path = (
    data_dir / f"labels_mapping_to_center_and_normalization/{image_number}.txt"
)
annotations = load_annotations(annotation_path)
# print(annotations[0])
# annotations = [annotations[0]]
result_image = draw_bboxes(image_path, annotations, OUTPUT_IMG_SIZE)
cv2.imwrite(str(output_path / f"{image_number}_complete.png"), result_image)
```
