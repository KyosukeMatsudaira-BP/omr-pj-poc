```python
import os
def modify_class_labels(annotation_dir, save_dir, class_mapping):
    """
    クラスラベルをマッピングに従って変更する関数

    Args:
        annotation_dir (str): アノテーションファイルのディレクトリ
        class_mapping (dict): 変更前と変更後のクラスラベルの対応辞書
        例: {0: 1, 2: 3} # クラス0→1、クラス2→3に変更
    """
    def modify_line(line):
        values = line.strip().split(',')

        # クラスラベルを変更（マッピングに存在する場合のみ）
        old_label = values[1]
        if old_label in class_mapping:
            values[1] = str(class_mapping[old_label])
        return ' '.join(values) + '\n'

    # ディレクトリ内の全てのtxtファイルを処理
    for filename in os.listdir(annotation_dir):
        if filename.endswith('.txt'):
            file_path = os.path.join(annotation_dir, filename)
            with open(file_path, 'r') as file:
                lines = file.readlines()

            save_file_path = os.path.join(save_dir, filename)
            # 修正された内容を書き込む
            with open(save_file_path, 'w') as file:
                modified_lines = [modify_line(line) for line in lines]
                file.writelines(modified_lines)
```

```python
# クラスマッピングを定義
class_mapping = {
    "_": 0,    # 音符以外の記号
    "-12": 1,  # 音符（C）
    "-11": 2,  # 音符（D）
    "-10": 3,  # 音符（E）
    "-9": 4,   # 音符（F）
    "-8": 5,   # 音符（G）
    "-7": 6,   # 音符（A）
    "-6": 7,   # 音符（B）
    "-5": 8,   # 音符（C）
    "-4": 9,   # 音符（D）
    "-3": 10,  # 音符（E）
    "-2": 11,  # 音符（F）
    "-1": 12,  # 音符（G）
    "0": 13,   # 音符（A）
    "1": 14,   # 音符（B）
    "2": 15,   # 音符（C）
    "3": 16,   # 音符（D）
    "4": 17,   # 音符（E）
    "5": 18,   # 音符（F）
    "6": 19,   # 音符（G）
    "7": 20,   # 音符（A）
    "8": 21,   # 音符（B）
    "9": 22,   # 音符（C）
    "10": 23,  # 音符（D）
    "11": 24,  # 音符（E）
    "12": 25,  # 音符（F）
}

# アノテーションディレクトリのパスを指定して実行
annotation_dir = 'data/dense/labels/'
save_dir = 'data/dense/labels_mapping/'
modify_class_labels(annotation_dir, save_dir, class_mapping)
```
